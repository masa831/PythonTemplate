```python
import numpy as np
from numba import njit, jit
from itertools import product
import time
import random
```

```python
shape = (10, 10, 10, 10, 10)
```

```python
%%timeit
# numpy
array_numpy = np.zeros(shape)
for i, j, k, l, m in product(range(10), repeat=5):
    array_numpy[i, j, k, l, m] = random.randint(0, 10)
# check
array_numpy[0, 0, 0, 0, :]
```

```python
@jit(nopython=True)
def calc_array(shape):
    arr = np.zeros(shape)
    for i in range(10):
        for j in range(10):
            for k in range(10):
                for l in range(10):
                    for m in range(10):
                        arr[i, j, k, l, m] = random.randint(0, 10)  # ランダム値を代入
    return arr
```

```python
%%timeit
arr = calc_array(shape)
# check
arr[0, 0, 0, 0, :]
```

```python
@njit
def calc_array(shape):
    arr = np.zeros(shape)
    for i in range(10):
        for j in range(10):
            for k in range(10):
                for l in range(10):
                    for m in range(10):
                        arr[i, j, k, l, m] = random.randint(0, 10)  # ランダム値を代入
    return arr
```

```python
%%timeit
arr = calc_array(shape)
# check
arr[0, 0, 0, 0, :]
```
