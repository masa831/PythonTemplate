appendix:　浅いコピーと深いコピー

```python
from copy import copy
```

```python
# コピー対象が一次元の場合は特に問題なし
org = [1,2,3,4,5]
cpy = org.copy()
```

```python
org[2] = 100
print(f'{org=}')
print(f'{cpy=}')
```

```python
cpy[3] = 10
print(f'{org=}')
print(f'{cpy=}')
```

```python
# コピー対象がlist[Union[list, dict]]などの場合には参照コピーとなる
org = [[1,2,3],[4,5]]
cpy = org.copy()
```

```python
org[0][2] = 100
print(f'{org=}')
print(f'{cpy=}')
```

```python
cpy[1][0] = 10
print(f'{org=}')
print(f'{cpy=}')
```

以下のようなデータクラスの場合でも、発生する。
List[Dataclass]の場合も、必要に応じて、deepcopyを用いること

```python
# test
@dataclass
class Test:
    id: int
    item: List

a = Test(1, [1,2,3])
b = Test(2, [11,21,31])
c = Test(3, [10,20,30])

l = [a, c, b]
cpyl = l.copy()
```

```python
l[0].item = [2,3,4]

print(f'{l=}')
print(f'{cpyl=}')
```
