```python
# 関数化　例 先頭を取得する
def get_key_from_value(d, val):
    keys = [k for k, v in d.items() if v == val]
    if keys:
        return keys[0]
    return None

key = get_key_from_value(d, 1200)
key = get_key_from_value(d, 2)
print(key)
```

Q 要素を削除する方法を列挙
* 指定したキーを削除
* 削除＆値を受け取り
* 削除＆キー、値を受け取り # 最後尾から削除

```python
d = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5}
del d['a']
print(d)

d = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5}
v = d.pop('a')
print(f'{v=}')
print(f'{d=}')

d = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5}
k, v = d.popitem()
print(f'{k=}')
print(f'{v=}')
print(f'{d=}')
```

Q 浅いコピー方法を三つ列挙

```python
d = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5}

d1 = dict(**d)
print(f'{d1}')

d2 = dict(d)
print(f'{d2}')

d3 = d.copy()
print(f'{d3}')
```

Q 辞書をマージする　※重複キーは上書き　2パターン  
* 新規生成
* どちらかの辞書を更新

```python
d1 = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5}
d2 = {'a': 1, 'b2': 2, 'c2': 3, 'd': 4, 'e': 10}

d = {**d1, **d2}
print(f'{d=}')

# eのように同じ値があった場合は、第二引数の方が優先される
dd = {**d2, **d1}
print(f'{dd=}')

d1.update(d2)
print(f'{d1=}')
```

Q 問い合わせ　キーの存在

```python
d = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5}
target = 'a'
if target in d:
    print(f'{target} in dict. val={d[target]}')
```

Q 問い合わせ　値の存在

```python
d = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5}
target = 2
if target in d.values():
    keys = [k for k,v in d.items() if v == target]
    print(f'{target} in dict. keys:{keys}')
```

Q 複数キーの存在判定　all, any

```python
d = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5}
target = {'a', 'b'}
chk = all(key in d for key in target)
print(chk)

target = {'a', 'q'}
chk = all(key in d for key in target)
print(chk)

target = {'a', 'q'}  # set変数
target = ['a', 'q']  # list変数どちらでもOK
chk = any(key in d for key in target)
print(chk)
```

Q 指定された複数のキーに対応する要素を抽出

```python
d = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5}
target = {'a', 'c'}

r = {}
for k, v in d.items():
    if k in target:
        r[k] = d[k]
print(f'{r=}')
```

Q 指定された複数のキーに対応する要素を抽出 内包表記バージョン

```python
d = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5}
target = {'a', 'c'}

r = {k:v for k, v in d.items() if k in target}
print(f'{r=}')
```

Q dataclassを辞書に変換する

```python
from typing import List
from dataclasses import dataclass, asdict

@dataclass
class Child:
  name: str

@dataclass
class Parent:
  name: str
  children: List[Child]

obj = Parent(name="bob", children=[Child(name="tom"), Child(name="mary")])
asdict(obj)
# => {'name': 'bob', 'children': [{'name': 'tom'}, {'name': 'mary'}]}
```
