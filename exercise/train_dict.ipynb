Q 値からキーを取得する

```python
d = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 2}
```

```python
# 関数化　例 先頭を取得する
def get_key_from_value(d, val):
    keys = [k for k, v in d.items() if v == val]
    if keys:
        return keys[0]
    return None

key = get_key_from_value(d, 1200)
key = get_key_from_value(d, 2)
print(key)
```

Q dataclassを辞書に変換する

```python
from typing import List
from dataclasses import dataclass, asdict

@dataclass
class Child:
  name: str

@dataclass
class Parent:
  name: str
  children: List[Child]

obj = Parent(name="bob", children=[Child(name="tom"), Child(name="mary")])
```
