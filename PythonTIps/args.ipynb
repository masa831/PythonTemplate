動作確認

```python
# sample: args
def func_args(a, b, *args):
    print(a)
    print(b)
    print(args)
```

```python
print(func_args(1,2,3,4,5))  # argsの部分はタプルとして入る
print()
print(func_args(1,2))
```

```python
def func_kwargs(a, b, **kwargs):
    print(a)
    print(b)
    print(kwargs)
```

```python
# これはエラー
# print(func_kwargs(1,2,3,4,5))  # argsの部分はタプルとして入る
print(func_kwargs(a=1, b=2, c=3, d=4, e=5))  # kwargsの部分は辞書として入る
print()
print(func_kwargs(1,2))
```

使用イメージ

```python
def func_args_sample(*args):
    print(args[0])
    print(len(args))
```

```python
print(func_args_sample(1))
print(func_args_sample(1,2,3,4,5))
```

```python
# kwargs
def func_kwargs_sample(**kwargs):
    print(kwargs.get('a'), kwargs.get('b'))
```

```python
print(func_kwargs_sample(a=1))
print(func_kwargs_sample(d=1))
print(func_kwargs_sample(a=1,b=2,c=3))
```
