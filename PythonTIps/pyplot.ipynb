```python
import matplotlib.pyplot as plt
```

```python
# 折れ線グラフ
plt.plot([1, 2, 3, 4], [10, 20, 25, 30], label="Line")
plt.legend()  # 凡例表示
plt.show()
```

```python
# 散布図
plt.scatter([1, 2, 3, 4], [10, 20, 25, 30], color="red", label="Points")
plt.legend()
plt.show()
```

```python
# 棒グラフ
plt.bar(["A", "B", "C"], [10, 20, 15], color="blue", label="Bars")
plt.legend()
plt.show()
```

```python
# ヒストグラム
data = [1, 1, 2, 2, 2, 3, 3, 4, 5, 5, 5, 5]
plt.hist(data, bins=5, color="green", alpha=0.7)
plt.show()
```

```python
# 円グラフ
labels = ["A", "B", "C"]
sizes = [30, 45, 25]
plt.pie(sizes, labels=labels, autopct="%1.1f%%", startangle=90)
plt.show()
```

```python
# customize
# sample data
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

x2 = [2, 3, 4, 5, 8]
y2 = [5, 20, 44, 30, 53]


plt.plot(x, y, color="blue", marker="o", linestyle="--", linewidth=2, markersize=6, label='L1')
# add grid
plt.grid(color="gray", linestyle="--", linewidth=0.5)
# axes
plt.xlim(0, 10)  # X軸範囲
plt.ylim(0, 50) # Y軸範囲
# log scale
# plt.yscale("log")

plt.show()
```

```python
# 重ねて表示
plt.plot(x, y, label="Line 1")
plt.plot(x2, y2, label="Line 2")
plt.legend()
plt.show()
```

```python
# 複数グラフをまとめ表示
fig, ax = plt.subplots(2, 1, figsize=(6, 8))  # 2行1列のサブプロット
ax[0].plot(x, y, label="Plot 1")
ax[1].plot(x2, y2, label="Plot 2")
plt.show()
```
