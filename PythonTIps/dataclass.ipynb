```python
from dataclasses import dataclass, field
from typing import List, Dict

@dataclass
class Person:
    name: str
    age: int
    address: str = 'tokyo'
    # list, dict, setを型指定なしで初期化する
    book : list = field(default_factory=list)
    y: dict = field(default_factory=dict)
    z: set = field(default_factory=set)
    # list, dict, setを型指定して、初期化する
    book2 : List[int] = field(default_factory=list)
    book3 : Dict[str, int] = field(default_factory=dict)
    #その他、fieldの活用方法
    active : bool = field(default=True, repr=False)  # repr=Falseとすると、printで表示されない
    info : str = field(init=False)

    # __post_init__を使うと初期化後に処理を追加できる
    def __post_init__(self):
        self.info = self.name + '/' + self.address
```

```python
taro = Person('taro', 30, 'tyokyo')
```

```python
# 引数が足りないとエラー
# hitosi = Person('hitoshi')
```

```python
ken = Person('ken', 30, 'osaka')

# データにアクセスする
print(f'{ken.name=}')
print(f'{ken.age=}')
print(f'{ken.address=}')
print(f'{ken.active=}')  # repr=Falseでも同様にアクセスできる
```

```python
# dict型へ変換できる
from dataclasses import asdict, astuple

print(asdict(ken))
```

```python
print(astuple(ken))
```

```python
# Frozen変数を使うことでイミュータブルにできる
@dataclass(frozen=True)
class homekey():
    no: int = 100
    password: str = "AAA"
```

```python
mykey = homekey()
mykey.no = 200  # Error
```

```python
# データクラスの内部を表示
from dataclasses import asdict
for k, v in asdict(ken).items():
    print(f'{k} \t: {v}')
```
