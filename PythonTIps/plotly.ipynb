```python
import plotly.express as px
import plotly.graph_objects as go
import pandas as pd
```

```python
# 折れ線グラフ
data = {
    "Year": [2018, 2019, 2020, 2021],
    "Sales": [200, 300, 400, 500],
}

# データフレーム作成
df = pd.DataFrame(data)
display(df)

# グラフ作成
fig = px.line(df, x="Year", y="Sales", title="Sales Over Years")
fig.show()
```

```python
# 散布図
data = {
    "x": [1, 2, 3, 4, 5],
    "y": [10, 20, 25, 30, 35],
    "group": ["A", "B", "A", "B", "A"],
}
df = pd.DataFrame(data)
display(df)

# グラフ作成
fig = px.scatter(df, x="x", y="y", color="group", title="Scatter Plot Example")
fig.show()
```

```python
# 棒グラフ
categories = ["Apples", "Oranges", "Bananas"]
values = [10, 15, 7]

fig = go.Figure(data=[go.Bar(x=categories, y=values)])
fig.update_layout(title="Bar Chart Example", xaxis_title="Fruits", yaxis_title="Quantity")
fig.show()
```

```python
# レイアウトカスタマイズ
fig.update_layout(
    # title
    title="カスタマイズされたグラフ",
    title_font={"size": 24, "color": "blue", "family": "Arial"},
    title_x=0,  # 0:left, 1:right, 0.5:middle
    title_y=0.9,  # 0: , 1:
    # axes
    xaxis_title="カテゴリ",
    yaxis_title="値",
    xaxis={
        "showgrid": True,       # グリッドを表示
        "zeroline": True,       # 原点の線を表示
        "title": "X軸の名前",    # 軸タイトル
        "title_font": {"size": 20, "color": "red"},  # 軸タイトルのフォント
        "tickangle": 45         # ラベルの角度
    },
    yaxis={
        "showline": True,       # 軸線を表示
        "ticks": "outside",     # 軸ラベルを外側に表示
        "range": [0, 100]       # Y軸の表示範囲を指定
    },
    # 背景、テーマ
    # template="plotly_dark",  # テーマ (例: 'plotly', 'seaborn', 'ggplot2')
    # plot_bgcolor="lightgray",  # plotエリアの背景
    # paper_bgcolor="white",  # 全体の背景
    # legend
    showlegend=False,  # 凡例の表示、非表示
    legend={
        "title": {"text": "凡例タイトル"},
        "x": 0.5,               # 凡例のX位置
        "y": 1.0,               # 凡例のY位置
        "font": {"size": 12},   # 凡例のフォントサイズ
        "bgcolor": "rgba(255,255,255,0.5)"  # 背景色 (透明度付き)
    },
    # マージン
    margin={"l": 50, "r": 50, "t": 80, "b": 50},
    # annotations
    annotations=[
        {
            "x": 2, "y": 30,    # 注釈の位置
            "xref": "x", "yref": "y",
            "text": "ここがポイント",  # テキスト
            "showarrow": True,   # 矢印を表示
            "arrowhead": 2,      # 矢印スタイル
            "font": {"size": 14}
        }
    ],
)

fig.show()
```

```python
df = pd.DataFrame([
    dict(Task="Job A", Start='2009-01-01', Finish='2009-02-28', Resource="Alex"),
    dict(Task="Job B", Start='2009-03-05', Finish='2009-04-15', Resource="Alex"),
    dict(Task="Job C", Start='2009-02-20', Finish='2009-05-30', Resource="Max")
])

fig = px.timeline(
    df,
    x_start="Start",
    x_end="Finish",
    y="Task",
    color="Resource",  # 色分けするカテゴリ列
    title="Gantt chart",
    hover_name="Resource",  # hoverで表示する列
    # hover_data=
    template='plotly',  # 'plotly_dark', 'ggplot2'
    # template='plotly_dark'
    # template='ggplot2'
    )
fig.update_yaxes(autorange="reversed")
fig.show()
```
