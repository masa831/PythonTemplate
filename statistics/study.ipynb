二項分布
成功するかしないかの二択の事象が起こる現象において、
成功確率をp、試行回数をnとすると成功回数が従う確率分布を二項分布という

例：
p=0.2, n=10のとき、3回成功する確率を求めることができる。


```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

plt.style.use("seaborn-darkgrid")

n = 10 #試行回数
p = 0.5 #確率
x = np.arange(n + 1)

plt.bar(x, stats.binom.pmf(x, n, p))
plt.xlabel('number of successes')
plt.ylabel('probability')
```

```python
# 確率の計算例
n = 100 #試行回数
p = 0.2 #確率
x = 3
prob = stats.binom.pmf(x, n, p)
prob
```

```python
# 正規分布
loc = 4  # 平均値
scale = 1  # 標準偏差
x = np.arange(0, 8, 0.1)

y = stats.norm.pdf(x, loc, scale)
plt.plot(x, y)
```
